```python
df = pd.read_csv("products.csv")
df.head()

# Kateqoriyaya görə orta qiymət
print(df.groupby("category")["price"].mean())

# Qiymətlərin histogramı
plt.hist(df["price"], bins=5, color="skyblue", edgecolor="black")
plt.title("Qiymətlərin paylanması")
plt.xlabel("Qiymət")
plt.ylabel("Miqdar")
plt.show()

# Rating-ə görə vizual
sns.barplot(x="category", y="rating", data=df)
plt.title("Kateqoriyalara görə orta reytinq")
plt.show()
```
